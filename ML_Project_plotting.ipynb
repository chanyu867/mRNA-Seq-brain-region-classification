# Brain-region classification: plots

The compressed input matrix is bundled with this repository; set `BRAIN_REGION_DATA_FILE` to use another file. Results live in `results/`; generated figures and tables can be inspected there.

Run this notebook after the analysis notebook. It reads saved scores and predictions and creates figures; it does not train models.


```python
# Project-wide paths: change DATA_FILE here or set BRAIN_REGION_DATA_FILE.
import os
from pathlib import Path

PROJECT_DIR = Path.cwd().resolve()
DATA_FILE = Path(os.environ.get("BRAIN_REGION_DATA_FILE",
                            PROJECT_DIR / "GSE211792_logCPM_cent.csv.gz")).expanduser().resolve()
RESULTS_DIR = PROJECT_DIR / "results"
RESUME_CHECKPOINT_DIR = PROJECT_DIR / "resume_checkpoints"

assert DATA_FILE.is_file(), "Input data missing; set BRAIN_REGION_DATA_FILE to a .csv or .csv.gz file."
RESULTS_DIR.mkdir(parents=True, exist_ok=True)
print("Input CSV found; result directory ready.")
```

```python
import pandas as pd
from IPython.display import Image, display
from project_plots import (create_pca_plots, create_performance_boxplots,
                           create_confusion_matrices)
```

## 1. Exploratory PCA

The same PC1–PC2 coordinates are colored by 25 regions, 4 macro regions, 7 macro regions, donor, and cell type. PCA is descriptive and uses all samples.

```python
pca_figures = create_pca_plots(PROJECT_DIR, DATA_FILE, RESULTS_DIR)
for path in pca_figures:
    print(path.relative_to(PROJECT_DIR))
    display(Image(filename=str(path)))
```

## 2. Donor-held-out performance

Each box summarizes macro F1 over six held-out donors. Macro-only and 25-region targets have different numbers of classes.

```python
performance_figures = create_performance_boxplots(PROJECT_DIR, RESULTS_DIR)
for path in performance_figures:
    print(path.relative_to(PROJECT_DIR))
    display(Image(filename=str(path)))
```

## 3. Confusion matrices

The direct, separate two-stage, joint, and macro-only matrices use saved held-out predictions. No model is refitted here.

```python
confusion_figures = create_confusion_matrices(PROJECT_DIR, RESULTS_DIR)
for path in confusion_figures:
    print(path.relative_to(PROJECT_DIR))
    display(Image(filename=str(path)))
```
